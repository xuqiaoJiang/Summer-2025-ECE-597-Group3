#BoW - MultinomialNB

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# load data
df = pd.read_csv('phishing_email.csv')

# extract features and target
X = df['text_combined']
y = df['label']

# Vectorization text using Bag-of-words
vectorizer = CountVectorizer(stop_words='english', max_df=0.7)
X_bow = vectorizer.fit_transform(X)

# split train & test dataset
X_train, X_test, y_train, y_test = train_test_split(X_bow, y, test_size=0.2, random_state=42)

# train model
model = MultinomialNB()
model.fit(X_train, y_train)

# predict and evaluable
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

# TF-IDF & MultinomiaNB

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# load data
df_2 = pd.read_csv("phishing_email.csv")

# extract features and target
X_2 = df_2["text_combined"]
y_2 = df_2["label"]

# Vectorization text using TF-IDF
vectorizer_2 = TfidfVectorizer(stop_words='english', max_df=0.7)
X_vec_2 = vectorizer_2.fit_transform(X_2)

# split train & test dataset
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(X_vec_2, y_2, test_size=0.2, random_state=42)

# train model
model_2 = MultinomialNB()
model_2.fit(X_train_2, y_train_2)

# predict and evaluable
y_pred_2 = model_2.predict(X_test_2)
print("Accuracy:", accuracy_score(y_test_2, y_pred_2))
print("Classification Report:\n", classification_report(y_test_2, y_pred_2))
```
